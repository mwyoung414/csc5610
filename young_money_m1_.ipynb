```python
def lin_reg_values(y: pd.Series, x: pd.Series) -> None:

    slope, intercept, r, p, stderr = stats.linregress(y, x)

    print(f'Linear Regression values for {y.name} vs {x.name}\n')
    print(f'Slope: {slope}')
    print(f'Intercept: {slope}')
    print(f'r {r}')
    print(f'p: {p}')
    print(f'stderr: {stderr}')
```

```python
lin_reg_values(used_cars_pd['Price'], used_cars_pd['Weight'])
```

```python
lin_reg_values(used_cars_pd['Price'], used_cars_pd['KM'])
```

```python
lin_reg_values(used_cars_pd['Price'], used_cars_pd['Age_08_04'])
```
