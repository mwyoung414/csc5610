```python
boolean_columns = [c for c in used_cars_pd.columns if used_cars_pd[c].dtype in ['bool']]
```

```python
def get_h_test_results(df: pl.DataFrame, group_col: str, value_col: str) -> Tuple[float, float]:
    samples_by_group = (
        df
        .group_by(group_col)
        .agg(pl.col(value_col))
    )
    
    s = samples_by_group[value_col].to_list()

    stat, p = stats.kruskal(*s)
    
    return stat, p
```

```python
for col in boolean_columns:
    stat, p = get_h_test_results(used_cars, col, 'Price')
    print(f'Kruskal-Wallis H-test results for {col} vs Price\n')
    print(f'stat: {stat}, p-value: {p}')
    print('-------------------------------------')
    print('\n')
```

```python
formula = 'Price ~ C(ABS) + Age_08_04 + C(Airbag_1) + C(Airbag_2) + C(Airco) + C(Automatic) + C(Automatic_airco) + C(Backseat_Divider) + C(Boardcomputer) + C(BOVAG_Guarantee) + CC + C(CD_Player) + C(Central_Lock) + C(Color) + C(Cylinders) + C(Doors) + C(Fuel_Type) + Gears + Guarantee_Period + HP + KM + C(Metallic_Rim) + C(Mfr_Guarantee) + C(Mistlamps) + C(Parking_Assistant) + C(Powered_Windows) + C(Power_Steering) + Quarterly_Tax + C(Radio) + C(Radio_cassette) + C(Tow_Bar) + Weight'
```

```python
# Using patsy to create design matrices for regression
y, X = dmatrices(formula, data=used_cars.to_pandas())
```

```python
used_car_model = sm.OLS(y, X).fit()
```

```python
summary = pd.DataFrame({
    'Features': used_car_model.model.exog_names,
    'P-values': used_car_model.pvalues,
})

summary.sort_values('P-values')
```

```python
for col in used_cars.columns:
    if used_cars[col].dtype in [pl.Float64, pl.Int64]:
        feature = f'Price ~ {col}'
        y_single, X_single = dmatrices(feature, data=used_cars.to_pandas())
        model = sm.OLS(y_single, X_single).fit()
        print(f'{col}: P-value: {model.pvalues[1]}, R-squared: {model.rsquared}, Adj. R-squared: {model.rsquared_adj}')
        print('-------------------------------------')
```

```python
bool_dict = {}
for col in used_cars.columns:
    if used_cars[col].dtype == pl.Boolean:
        feature = f'Price ~ C({col})'
        y_single, X_single = dmatrices(feature, data=used_cars.to_pandas())
        model = sm.OLS(y_single, X_single).fit()
        bool_dict[col] = {
            "P-value": model.pvalues[1],
            "R-squared": model.rsquared,
            "Adj. R-squared": model.rsquared_adj
        }

sorted_by_p = sorted(bool_dict.items(), key=lambda kv: kv[1].get("P-value", float("inf")))
for feature, stats in sorted_by_p:
    print(feature, "P-value:", stats.get("P-value"), "R-squared:", stats.get("R-squared"))
```
