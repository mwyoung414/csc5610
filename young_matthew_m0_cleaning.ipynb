####  Convert Types

Use the DataFrame astype() method to convert columns to the correct type and overwrite the original column in the DataFrame.  Confirm that all of the columns are converted to the desired types using the info() method.  If you get an error, you may have missed a bad value and should look at the values in the column more closely.




```python
# Identify columns with "Yes"/"No" values and saves to a a list

## Automatic column will not be included since it has NaN values

yes_no_cols = []

for col in used_cars_pl_df.columns:
    if used_cars_pl_df[col].dtype == pl.Utf8:
        uniques = [v.lower() for v in used_cars_pl_df[col].unique().to_list() if v is not None]
        if set(uniques).issubset({"yes", "no"}):
            yes_no_cols.append(col)
```

```python
print(yes_no_cols)
```

```python
# Convert "Yes"/"No" columns to boolean True/False, else None

used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.when(pl.col(col).str.to_lowercase() == "yes").then(True)
    .when(pl.col(col).str.to_lowercase() == "no").then(False)
    .otherwise(None)
    .alias(col)
    for col in yes_no_cols
)
```

```python
# Convert Mfg_Date to datetime format

used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.col("Mfg_Date").str.strptime(pl.Date, "%m/%d/%y").alias("Mfg_Date")
)
```

```python
# Convert the Categorical Fuel_Type column to categorical data type
used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.col("Fuel_Type").cast(pl.Categorical).alias("Fuel_Type")
)
```

```python
# Convert the Categorical Color column to categorical data type
used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.col("Color").cast(pl.Categorical).alias("Color")
)
```

```python
# Decided to convert Automatic column to boolean True/False, with nan the column would have to to be a string/object data type

used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.when(pl.col("Automatic") == "Yes").then(True)
    .otherwise(False)
    .alias("Automatic")
    .cast(pl.Boolean)
)
```

```python
used_cars_pl_df = used_cars_pl_df.with_columns(
    used_cars_pl_df['Model'].str.replace(r"TOYOTA Corolla\s*", "").alias("Model")
)
```

```python
used_cars_pl_df = used_cars_pl_df.with_columns(
    used_cars_pl_df['Model'].str.replace(r"\b\d\.\d\b", "").alias("Model")
)
```

```python
### Remove door count designations from Model column

used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.col("Model")
      # 2/3-Doors or 4/5-Doors (hyphen optional, case-insensitive)
      .str.replace_all(r"(?i)\b\d+/\d+\s*(?:-\s*)?doors?\b", "", literal=False)
      # 3-Drs, 3 - Drs
      .str.replace_all(r"(?i)\b\d+\s*-\s*drs\b", "", literal=False)
      # 3DR, 3DRS (no hyphen)
      .str.replace_all(r"(?i)\b\d+\s*drs?\b", "", literal=False)
      # tidy spaces after removals
      .str.replace_all(r"\s+", " ", literal=False)
      .str.strip_chars()
      .alias("Model")          # overwrite original column
)
```

```python
# add Valve to New Column

used_cars_pl_df = used_cars_pl_df.with_columns([
    # Text tag like "16V" (handles optional space: "16 V"), null if not present
    pl.col("Model")
      .str.extract(r"(?i)\b(\d+\s*V)\b", 1)   # first match only
      .str.replace_all(r"\s+", "")            # normalize "16 V" -> "16V"
      .str.to_uppercase()
      .alias("Valve"),

    # Optional: numeric total as an integer (e.g., 16)
    pl.col("Model")
      .str.extract(r"(?i)\b(\d+)\s*V\b", 1)
      .cast(pl.Int16)
      .alias("Valve_Total")
])
```

```python
used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.col("Model")
      .str.replace_all(r"(?i)\b\d+\s*V\b", "", literal=False)
      .str.replace_all(r"(?i)\b\d+\s*valves?\b", "", literal=False)
      .str.replace_all(r"\s+", " ", literal=False)
      .str.strip_chars()
      .alias("Model")
)
```

```python
trim_expr = (
    pl.when(pl.col("Model").str.contains(r"(?i)\bLinea\s+Terra\b", literal=False)).then(pl.lit("Linea Terra"))
     .when(pl.col("Model").str.contains(r"(?i)\bLinea\s+Luna\b",  literal=False)).then(pl.lit("Linea Luna"))
     .when(pl.col("Model").str.contains(r"(?i)\bLinea\s+Sol\b",   literal=False)).then(pl.lit("Linea Sol"))
     .when(pl.col("Model").str.contains(r"(?i)\bT[ -]?Sport\b",   literal=False)).then(pl.lit("T Sport"))
     .when(pl.col("Model").str.contains(r"(?i)\bExec(?:\s+BNS)?\b", literal=False)).then(pl.lit("Executive"))
     .when(pl.col("Model").str.contains(r"(?i)\bG6-R\b",          literal=False)).then(pl.lit("G6-R"))
     .when(pl.col("Model").str.contains(r"(?i)\bG6\b",            literal=False)).then(pl.lit("G6"))
     .when(pl.col("Model").str.contains(r"(?i)\bTerra\b",         literal=False)).then(pl.lit("Terra"))
     .when(pl.col("Model").str.contains(r"(?i)\bLuna\b",          literal=False)).then(pl.lit("Luna"))
     .when(pl.col("Model").str.contains(r"(?i)\bSol\b",           literal=False)).then(pl.lit("Sol"))
     .otherwise(pl.lit(None))
)

used_cars_pl_df = used_cars_pl_df.with_columns(
    trim_expr.cast(pl.Categorical).alias("Trim")
)
```

```python
used_cars_pl_df = used_cars_pl_df.with_columns([
    # Body_Style: add Wagon/Stationwagon/Stationwagen detection
    pl.when(pl.col("Model").str.contains(r"(?i)\b(Stationwag(?:on|en)|Wagon|WGN)\b", literal=False))
      .then(pl.lit("Wagon"))
     .when(pl.col("Model").str.contains(r"(?i)\b(HATCHB|HB|HATCHBACK|LIFTB|LIFTBACK)\b", literal=False))
      .then(pl.lit("Hatchback"))
     .when(pl.col("Model").str.contains(r"(?i)\b(SEDAN|SDN)\b", literal=False))
      .then(pl.lit("Sedan"))
     # (optional) MPV:
     .when(pl.col("Model").str.contains(r"(?i)\bMPV\b", literal=False))
      .then(pl.lit("MPV"))
     .otherwise(pl.lit(None))
     .alias("Body_Style")
])
```

```python
used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.when(pl.col("Model").str.contains(r"(?i)\bDual\s+VVT[-\s]?i\b", literal=False)).then(pl.lit("Dual VVT-i"))
     .when(pl.col("Model").str.contains(r"(?i)\bVVTL[-\s]?i\b", literal=False)).then(pl.lit("VVTL-i"))
     .when(pl.col("Model").str.contains(r"(?i)\bVVT[-\s]?i\b", literal=False)).then(pl.lit("VVT-i"))
     .otherwise(pl.lit(None))
     .alias("Valve_Tech")
)
```
