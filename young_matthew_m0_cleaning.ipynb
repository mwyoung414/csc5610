#### 3 - Handle Mixed Type and Missing Values that Prevent Type Conversions

```python
# Missing value with "?" in Automatic column, replaced with Nan

used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.when(pl.col("Automatic") == "?")
    .then(float("nan"))
    .otherwise(pl.col("Automatic"))
    .alias("Automatic")
)
```

```python
### All values are 4 except for a -1, changed -1 to Nan

used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.when(pl.col("Cylinders") == -1)
    .then(float('nan'))
    .otherwise(pl.col("Cylinders"))
    .alias("Cylinders")
)
```

```python
# Identify columns with "Yes"/"No" values and saves to a a list

## Automatic column will not be included since it has NaN values

yes_no_cols = []

for col in used_cars_pl_df.columns:
    if used_cars_pl_df[col].dtype == pl.Utf8:
        uniques = [v.lower() for v in used_cars_pl_df[col].unique().to_list() if v is not None]
        if set(uniques).issubset({"yes", "no"}):
            yes_no_cols.append(col)
```

```python
print(yes_no_cols)
```

```python
# Convert "Yes"/"No" columns to binary 1/0

used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.when(pl.col(col).str.to_lowercase() == "yes").then(True)
    .when(pl.col(col).str.to_lowercase() == "no").then(False)
    .otherwise(None)
    .alias(col)
    for col in yes_no_cols
)
```

```python
# Parse manufacturing date; pass the format as a positional argument (no 'fmt' keyword).
# The Mfg_Date values look like '10/1/02' so use '%m/%d/%y' and parse to a Date.
used_cars_pl_df = used_cars_pl_df.with_columns(
    pl.col("Mfg_Date").str.strptime(pl.Date, "%m/%d/%y").alias("Mfg_Date")
)
```

```python
# Aggregate by year to see how many cars were manufactured each year

data_hist = (used_cars_pl_df.group_by(
    pl.col("Mfg_Date").dt.year().alias("Year"),
    ).agg(pl.len()).sort("Year").to_pandas())

print(data_hist)
```

```python
# Histogram of cars manufactured by year

sns.histplot(data=data_hist, x='Year', weights="len", discrete=True, binwidth=1, stat='count')
xlabel = 'Year'
ylabel = 'Count'
plt.xlabel(xlabel)
plt.ylabel(ylabel)
plt.grid(axis='y', color='black', linewidth=0.7, linestyle='--', alpha=0.9)
plt.xlim(1997.5, 2004.5)
plt.show()
```
